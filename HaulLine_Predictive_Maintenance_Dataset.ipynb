# HaulLine Logistics — Predictive Maintenance Dataset

This notebook loads the **jskswamy/predictive-maintenance-data** dataset from Hugging Face.

Dataset source: https://huggingface.co/datasets/jskswamy/predictive-maintenance-data

The dataset is a tabular predictive-maintenance dataset containing engine sensor measurements and the `Engine Condition` target.

```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully.')
```

```python
# Load the complete dataset from Hugging Face
dataset = load_dataset('jskswamy/predictive-maintenance-data')

print(dataset)
```

```python
# Convert each split to a Pandas DataFrame
train_df = dataset['train'].to_pandas()
val_df = dataset['validation'].to_pandas()
test_df = dataset['test'].to_pandas()

print('Train shape:', train_df.shape)
print('Validation shape:', val_df.shape)
print('Test shape:', test_df.shape)
```

```python
# Combine all splits to get the complete dataset
df = pd.concat([train_df, val_df, test_df], ignore_index=True)

print('Complete dataset shape:', df.shape)
df.head(10)
```

```python
# Display all column names
print('Columns:')
for i, column in enumerate(df.columns, start=1):
    print(f'{i}. {column}')
```

```python
# Dataset information
df.info()
```

```python
# Statistical summary
df.describe().T
```

```python
# Check missing values
missing = df.isnull().sum()
print(missing[missing > 0] if missing.any() else 'No missing values found.')
```

```python
# Target distribution
print(df['Engine Condition'].value_counts())

df['Engine Condition'].value_counts().plot(kind='bar')
plt.title('Engine Condition Distribution')
plt.xlabel('Engine Condition (0 = Normal, 1 = Needs Maintenance)')
plt.ylabel('Number of Readings')
plt.show()
```

```python
# Example engine sensor distributions
sensor_columns = [
    'Engine RPM',
    'Lub Oil Pressure',
    'Fuel Pressure',
    'Coolant Pressure',
    'Lub Oil Temp',
    'Coolant Temp'
]

df[sensor_columns].hist(figsize=(14, 10), bins=30)
plt.suptitle('Engine Sensor Distributions')
plt.tight_layout()
plt.show()
```

## Dataset fields

The dataset includes engine measurements such as RPM, oil/fuel/coolant pressures, oil/coolant temperatures, engineered interaction features, and the `Engine Condition` target.

`Engine Condition` is the target used for predictive-maintenance classification.

```python
# Save the complete dataset locally as CSV (optional)
df.to_csv('haul_line_predictive_maintenance.csv', index=False)
print('Saved as haul_line_predictive_maintenance.csv')
```
